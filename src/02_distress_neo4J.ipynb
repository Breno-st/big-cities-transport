```python
import pandas as pd



path = "C:/buildbr/big-cities-transport/02.Distress/01.Input/"
mode = 'tube'
df = pd.read_csv(path+'od_'+mode+'.csv')
df['o_d'] = df.apply (lambda row: row['mode']+'_'+row['day']+'_'+row['name_o']+'_'+row['name_d'], axis=1)
df.set_index(['o_d'], inplace=True) # set o_d index
df = df.drop(['mode', 'day', 'name_o', 'name_d'], axis=1) # set o_d index
# Initialize an empty dictionary
result_dict = {}

for index, row in df.iterrows():
    # Create a sub-dictionary for each row
    row_data = {}    
    # Loop through the columns and add them as attributes
    for column in df.columns:
        if column != 'index_col':  # Skip the index column if needed
            row_data[column] = row[column]
    
    # Add the sub-dictionary to the main dictionary with the index as the key
    result_dict[index] = row_data


# result_dict will contain the desired dictionary
print(result_dict)
```
