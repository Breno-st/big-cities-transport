```python
def spearman_rank_correlation(rank_seq1, rank_seq2):
    """
    Calculate the Spearman Rank Correlation Coefficient between two rank sequences.

    Parameters:
    rank_seq1 (list): The first rank sequence.
    rank_seq2 (list): The second rank sequence.

    Returns:
    float: The Spearman Rank Correlation Coefficient, ranging from -1 to 1.
    """
    if len(rank_seq1) != len(rank_seq2):
        raise ValueError("Both rank sequences must have the same length")

    n = len(rank_seq1)

    # Calculate the rank differences squared
    d_squared = [(rank_seq1[i] - rank_seq2[i]) ** 2 for i in range(n)]

    # Calculate the Spearman Rank Correlation Coefficient
    r = 1 - (6 * sum(d_squared)) / (n * (n ** 2 - 1))

    return r
```

```python
import pandas as pd
from scipy.stats import spearmanr

# Step 1: Read Data from CSV
df = pd.read_csv('cardio_base.csv')

# Step 2: Calculate Spearman Rank Correlation for specific columns
columns_to_compare = ['height', 'weight', 'ap_hi', 'ap_lo', 'cholesterol', 'smoke']

for i in range(len(columns_to_compare)):
    for j in range(i + 1, len(columns_to_compare)):
        column1_data = df[columns_to_compare[i]]
        column2_data = df[columns_to_compare[j]]

        corr, _ = spearmanr(column1_data, column2_data)

        print(f"Spearman Rank Correlation between '{columns_to_compare[i]}' and '{columns_to_compare[j]}': {corr}")
```

```python
mean_height = df['height'].mean()
std_height = df['height'].std()

# Step 3: Determine the range within 2 standard deviations
lower_bound = mean_height - 2 * std_height
upper_bound = mean_height + 2 * std_height

# Step 4: Count the number of people outside this range
outliers = df[(df['height'] < lower_bound) | (df['height'] > upper_bound)]

# Step 5: Calculate the percentage of people outside this range
percentage_outliers = (len(outliers) / len(df)) * 100

print(f"The percentage of people 2 standard deviations away from the average height is: {percentage_outliers:.2f}%")
```

```python
import pandas as pd
from scipy.stats import ttest_ind

df = pd.read_csv('cardio_base.csv') 

# Separate data into smokers and non-smokers
smokers = df[df['smoke'] == 1]
non_smokers = df[df['smoke'] == 0]

# Calculate the means of weight for smokers and non-smokers
mean_weight_smokers = smokers['weight'].mean()
mean_weight_non_smokers = non_smokers['weight'].mean()

# Perform a two-sample t-test
t_stat, p_value = ttest_ind(smokers['weight'], non_smokers['weight'], equal_var=False)  # Assuming unequal variances

# Set your significance level (α)
alpha = 0.05

# Compare p-value to the significance level
if p_value < alpha:
    print("The mean weight of smokers is significantly different from non-smokers.")
else:
    print("There is no significant difference in weight between smokers and non-smokers.")

print(f"Mean weight of smokers: {mean_weight_smokers:.2f} kg")
print(f"Mean weight of non-smokers: {mean_weight_non_smokers:.2f} kg")
```

```python
periods = ['Morning', 'AM Peak', 'Midday', 'PM Peak', 'Evening', 'Late']

fri_matrix = [[0] * len(periods) for _ in range(len(periods))]

for i, period1 in enumerate(periods):
	for j, period2 in enumerate(periods):

		list1 = df.loc[:, ('fri',period1,'efficiencies_rnk')].values
		list2 = df.loc[:, ('fri',period2,'efficiencies_rnk')].values

		fri_matrix[i][j] = spearman_rank_correlation(list1, list2)
		

fri_matrix = np.array(fri_matrix)
```
